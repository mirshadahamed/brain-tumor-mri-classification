# Inspect the dataset

Check image counts, dimensions, color modes, and unreadable files. Images are not modified.

In VS Code, select the repository `.venv` Python kernel and run cells from top to bottom. The shared src/data_pipeline.py helper is retained for model.ipynb.


```python
from pathlib import Path
import sys

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "dataset").is_dir() and (p / "src").is_dir()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook inside the project repository.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("Repository:", ROOT)
```

```python
from collections import Counter

```

```python
from pathlib import Path

```

```python
from PIL import Image

```

```python
DATASET = ROOT / "dataset"

```

```python
CLASSES = ["glioma", "meningioma", "notumor", "pituitary"]

```

```python
EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

```

```python
total = 0

```

```python
problems = []

```

```python
for split in ["Training", "Testing"]:
    print(f"\n{split}")

    for class_name in CLASSES:
        folder = DATASET / split / class_name

        if not folder.is_dir():
            raise FileNotFoundError(f"Missing folder: {folder}")

        files = sorted(
            path for path in folder.iterdir()
            if path.is_file() and path.suffix.lower() in EXTENSIONS
        )

        sizes = Counter()
        modes = Counter()

        for path in files:
            try:
                with Image.open(path) as image:
                    image.load()
                    sizes[image.size] += 1
                    modes[image.mode] += 1
            except Exception as error:
                problems.append((str(path), str(error)))

        total += len(files)
        print(f"  {class_name}: {len(files)} images")
        print(f"    Common sizes: {sizes.most_common(3)}")
        print(f"    Color modes: {dict(modes)}")

```

```python
print(f"\nTotal images: {total}")

```

```python
print(f"Unreadable images: {len(problems)}")

```

```python
for path, error in problems:
    print(f"  {path}: {error}")
```
