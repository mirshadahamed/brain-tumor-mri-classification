# Predict a new MRI image

Place an image in `inputs/`, then set IMAGE_PATH below. Uses the frozen best CNN checkpoint. Probabilities are model scores, not calibrated clinical confidence. This assignment demo is not a diagnostic tool; the model will assign a class even to unsuitable input. Input images remain local and are ignored by Git.


```python
from pathlib import Path
import sys

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "dataset").is_dir() and (p / "src").is_dir()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook inside the project repository.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("Repository:", ROOT)
```

```python
IMAGE_PATH = ROOT / "inputs" / "download.jpeg"
```

```python
from src.prediction import predict_image
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

if not IMAGE_PATH.is_file():
    print(f"Add an MRI image and update IMAGE_PATH: {IMAGE_PATH}")
else:
    result = predict_image(IMAGE_PATH)
    print("Predicted class:", result["predicted_class"])
    scores = pd.Series(result["probabilities"], name="Probability")
    display((scores * 100).rename("Probability (%)").to_frame())
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))
    axes[0].imshow(result["image"])
    axes[0].set_title(f"Prediction: {result["predicted_class"]}")
    axes[0].axis("off")
    scores.plot.bar(ax=axes[1])
    axes[1].set(ylim=(0, 1), ylabel="Model probability")
    axes[1].tick_params(axis="x", rotation=30)
    fig.tight_layout()
    plt.show()
```
